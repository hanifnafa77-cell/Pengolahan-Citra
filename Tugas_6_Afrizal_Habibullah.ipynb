# **Jawaban Pratikum T6**

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data


image = data.camera()


plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Citra Grayscale')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.hist(image.ravel(), bins=256, range=(0, 256))
plt.title('Histogram Citra')
plt.xlabel('Intensitas Piksel')
plt.ylabel('Jumlah Piksel')

plt.tight_layout()
plt.show()
```

1. Citra yang digunakan merupakan gambar grayscale dengan nilai intensitas 0 sampai 255. Histogram menunjukkan persebaran jumlah piksel berdasarkan tingkat kecerahan. Bagian histogram yang berada di kiri menunjukkan piksel gelap, sedangkan bagian kanan menunjukkan piksel terang.

```python
brightness_values = [-50, 0, 50]

for b in brightness_values:

    result = np.clip(
        image.astype(np.int16) + b,
        0,
        255
    ).astype(np.uint8)

    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(result, cmap='gray')
    plt.title(f'Brightness {b}')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.hist(result.ravel(), bins=256, range=(0, 256))
    plt.title(f'Histogram Brightness {b}')
    plt.xlabel('Intensitas')
    plt.ylabel('Jumlah Piksel')

    plt.show()

plt.subplot(1, 2, 2)
plt.hist(image.ravel(), bins=256, range=(0, 256))
plt.title('Histogram Citra')
plt.xlabel('Intensitas Piksel')
plt.ylabel('Jumlah Piksel')

plt.tight_layout()
plt.show()
```

2. Ketika brightness ditambah, citra menjadi lebih terang dan histogram cenderung bergeser ke kanan. Sebaliknya, ketika brightness dikurangi, citra menjadi lebih gelap dan histogram cenderung bergeser ke kiri. Jika perubahan terlalu besar, sebagian nilai piksel dapat mencapai batas 0 atau 255

```python
contrast_values = [0.5, 1.0, 1.5]

for c in contrast_values:

    result = np.clip(
        128 + (image.astype(np.float32) - 128) * c,
        0,
        255
    ).astype(np.uint8)

    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(result, cmap='gray')
    plt.title(f'Contrast {c}')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.hist(result.ravel(), bins=256, range=(0, 256))
    plt.title(f'Histogram Contrast {c}')
    plt.xlabel('Intensitas')
    plt.ylabel('Jumlah Piksel')

    plt.show()
```

3. Pada contrast 0.5, perbedaan antara bagian gelap dan terang menjadi lebih kecil sehingga gambar terlihat lebih datar. Pada contrast 1.0, gambar tidak mengalami perubahan kontras. Pada contrast 1.5, perbedaan intensitas menjadi lebih besar sehingga gambar terlihat lebih tegas. Histogram juga mengalami perubahan persebaran intensitas.

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, exposure


image = data.camera()


low = np.percentile(image, 2)
high = np.percentile(image, 98)


stretched = exposure.rescale_intensity(
    image,
    in_range=(low, high),
    out_range=(0, 255)
).astype(np.uint8)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Citra Sebelum Contrast Stretching')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.hist(image.ravel(), bins=256, range=(0, 256))
plt.title('Histogram Sebelum')

plt.subplot(2, 2, 3)
plt.imshow(stretched, cmap='gray')
plt.title('Citra Setelah Contrast Stretching')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.hist(stretched.ravel(), bins=256, range=(0, 256))
plt.title('Histogram Setelah')

plt.tight_layout()
plt.show()
```

4. Setelah contrast stretching dilakukan, rentang intensitas piksel diperluas sehingga perbedaan antara bagian gelap dan terang menjadi lebih jelas. Gambar dapat terlihat lebih tajam dan detail lebih mudah diamati. Histogram juga menunjukkan persebaran intensitas yang lebih luas.

5. Kesimpulan
#
1. Perbedaan fungsi brightness dan contrast

Brightness berfungsi mengatur tingkat kecerahan gambar, sedangkan contrast berfungsi mengatur perbedaan intensitas antara bagian gelap dan terang pada gambar.

2. Pengaruh perubahan brightness dan contrast terhadap histogram

Perubahan brightness menyebabkan histogram bergeser ke kiri atau kanan. Sementara itu, perubahan contrast menyebabkan persebaran intensitas menjadi lebih sempit atau lebih luas.

3. Manfaat contrast stretching dalam meningkatkan kualitas citra

Contrast stretching bermanfaat untuk memperluas rentang intensitas gambar sehingga bagian gelap dan terang lebih mudah dibedakan. Dengan demikian, detail gambar dapat terlihat lebih jelas.
